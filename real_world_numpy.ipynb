```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("datasets/daraz_dataset.csv")
df.head()
df.shape
df.columns
df.dtypes
df["Age"]
df["Age"] = pd.to_numeric(df["Age"],errors="coerce")

numeric_columns = [
    "Age",
    "Quantity",
    "UnitPrice",
    "DiscountPercent",
    "TotalAmount",
    "Rating"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column],errors="coerce")

df.isnull().sum()
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].mean())

features = ["Age","Quantity","UnitPrice","DiscountPercent"]
target = "TotalAmount"

#generally freatures stores in x and target stores in y
X = df[features]
y = df[target]

type(X)
type(y)

X = X.to_numpy()
y = y.to_numpy()

type(X)
type(y)

# x
# X.shape
# X.ndim
# X.dtype
X[:,3]
print(X[X[:,0]>30])

X.dtype
```
