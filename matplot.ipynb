```python
dd = pd.read_csv("datasets/daraz_dataset.csv")
```

```python
category = dd.groupby("Category")["Product"].count()
category
```

```python
plt.figure(figsize=(18,6))
plt.bar(category.index, category.values,color="green",edgecolor="blue")
plt.xlabel("Categories",fontsize=20)
plt.ylabel("Products",fontsize=20)
plt.title("Product & Category Bar Graph")
plt.grid(axis="y",linestyle="--", alpha=1)
```

```python
category = dd.groupby("Category")["Product"].count()
category
```

```python
colors = [
    "red",
    "blue",
    "green",
    "grey",
    "black",
    "purple",
    "yellow"
]
plt.title("Product Distribution by category")
plt.pie(category.values,labels=category.index, colors=colors, shadow=True, startangle=90,explode=(0.1,0,0,0,0,0,0.1))
plt.figure(figsize=(14,6))
```

```python
plt.hist(dd["UnitPrice"],color="green", edgecolor="black",alpha=0.5)
plt.title("Product Distribution of UnitPrice")
plt.xlabel("UnitPrice")
plt.ylabel("Frequency")
```

```python
bins = [0,5000,10000,20000,50000,80000]
plt.hist(dd["UnitPrice"],bins=bins,color="green", edgecolor="black",alpha=0.5)
plt.title("Product Distribution of UnitPrice")
plt.xlabel("UnitPrice")
plt.ylabel("Frequency")
```

```python
plt.scatter(dd["UnitPrice"],dd["Rating"])
plt.xlabel("UnitPrice")
plt.ylabel("Rating")
plt.grid(True)
```

```python
data = {
    "Age" : [20,22,24,26,28,30],
    "TotalAmount" : [1000,1200,1500,2000,3000,32000]
}
dfd = pd.DataFrame(data)
plt.scatter(dfd["Age"],dfd["TotalAmount"])
plt.xlabel("Age")
plt.ylabel("TotalAmount")
plt.grid(True)
plt.title("Age vs TotalAmount")
```
