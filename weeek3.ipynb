```python
#2.Data Cleaning
#remove duplicates
df = df.drop_duplicates()
```

```python
#handle missing value
df['attendance_percent'] = df['attendance_percent'].fillna(df['attendance_percent'].median())
#Convert incorrect datatype into approptiate format
df['enrollment_date'] = pd.to_datetime(df['enrollment_date'])
```

```python
#Standardize inconsistant value
df['city'] = df['city'].replace({'ktm': 'Kathmandu'})
df['course'] = df['course'].replace({'Pythn AI': 'Python with AI/ML'})
```

```python
#Fix negative values
df['course_fee'] = df['course_fee'].abs()
df['assignment_score'] = df['assignment_score'].abs()
```

```python
# 3. Outlier detection using IQR
cols = ['course_fee', 'attendance_percent', 'assignment_score', 'final_score']
q1 = df[cols].quantile(0.25)
q3 = df[cols].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
iqr_summary = pd.DataFrame({
    'Q1': q1,
    'Q3': q3,
    'IQR': iqr,
    'Lower Bound': lower_bound,
    'Upper Bound': upper_bound
})
print(iqr_summary)
```

```python
#Outlier Handeling
outlier = ((df[cols] < lower_bound) | (df[cols] > upper_bound)).any(axis=1)
df_remove = df[~outlier]
print(df_remove)
```

```python
#replace outliers with median
median = df[cols].median()
df[cols] = df[cols].where((df[cols] >= lower_bound) & (df[cols] <= upper_bound),median,axis=1)
print(df)

```

```python
#windorization using clip()
df[cols] = df[cols].clip(lower=lower_bound, upper=upper_bound, axis=1)
print(df[cols].describe())
```

```python
#z_score
z_scores = (df[cols] - df[cols].mean()) / df[cols].std()
z_outlier = (z_scores > 3) | (z_scores < -3)
outliers_df = df[z_outlier.any(axis=1)]
print(outliers_df)
```
