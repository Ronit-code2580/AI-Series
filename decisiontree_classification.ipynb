```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv("datasets/student_multiclass_multiple_features.csv")

df["Result"]= df["Result"].map({
    "Fail" : 0,
    "Average" : 1,
    "Good" : 2
})
X = df[["StudyHours","Attendance","PreviousMarks"]]
y = df["Result"]

model = DecisionTreeClassifier(
    criterion="gini",
    random_state = 42
)
model.fit(X,y)
print(model.tree_.threshold)
new_student = pd.DataFrame({
    "StudyHours" : [50],
    "Attendance" : [95],
    "PreviousMarks": [90]
})
prediction = model.predict(new_student)
print(prediction)
```

```python
X = df[["StudyHours","EnglishScore","MathScore"]]
y = df[["PassEnglish","PassMath"]]
model = DecisionTreeClassifier(
    criterion="gini",
    random_state = 42
)
model.fit(X,y)
new_student = pd.DataFrame({
    "StudyHours" : [3],
    "EnglishScore" : [30],
    "MathScore": [95]
})
prediction = model.predict(new_student)
prediction
```
